```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
```

```python
data = pd.read_csv("/content/moons_demo.csv")
data.head()
data
```

```python
plt.scatter(data["x"], data["y"])
#
```

```python
model=DBSCAN(eps=0.1,min_samples=6)
labs = model.fit_predict(data)
print (f"number of clusters: {len(set(labs))}, noise: {(labs==-1).sum()}")
plt.scatter(data.x, data.y, c=labs)
```
